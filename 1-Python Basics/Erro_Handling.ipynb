```python
try:

    f= open('mytext.txt','r')

    f.write('This text will be added to the file mytext')
except OSError:
    print('This gives an OSError')
except IOError:
    print('This gives IO Error')
finally:
    print('I print anyways')
```

```python
try:

    f= open('mytext.txt','w')
    f.write('This text will be added to the file mytext')

    with open('mytext.txt','r') as file:
        contents=file.read()
        print(contents)
except OSError:
    print('This gives an OSError')
except IOError:
    print('This gives IO Error')
finally:
    print('I print anyways')
```
