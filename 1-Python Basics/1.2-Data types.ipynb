Integers --> int --> whole numbers --> 3,5,600
Floating numbers --> float --> 3.0 5.0 600.0
Strings --> str --> (Immutable) Ordered sequence of characters --> "Karthik", "Kalle"
Lists --> list --> Ordered sequence of objects --> [10,"karthik",5.0,{"a","b"}]
Tuples --> tup --> Ordered immutable sequence of objects --> (10,"Karthik",5.0)
Dictionary --> dict --> Unordered Key:Value Pairs --> {"Name":"Karthik", "age":32}
Sets --> set --> Unordered, Unique collection of objects --> {"a","b"}
Boolean --> bool --> Logical value indicating True or False
