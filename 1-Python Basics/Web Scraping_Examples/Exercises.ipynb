```python
two_star_titles = []

for n in range(1,51):
    scrape_url = base_url.format(n)

    res = requests.get(scrape_url)

    soap = bs4.BeautifulSoup(res.text,"html.parser")

    books = soap.select('.product_pod')

    for book in books:
        if [] == book.select('.star-rating.Two'):
            title = book.select("a")[1]['title']
            two_star_titles.append(title)

    
```

```python
authors = set()

page_valid =True

page_num = 1

org_url = 'https://quotes.toscrape.com/page/'

while page_valid:

    page_url = (org_url+str(page_num))

    reqs = requests.get(page_url)

    if "No quotes found!" in reqs.text:
        break

    beauty = bs4.BeautifulSoup(reqs.text,"html.parser")

    for author in beauty.select('.author'):
        authors.add(author.getText())

    page_num+=1


authors
```
