```python
phone_pattern = re.compile(r'(\d{5})-(\d{4})-(\d{2})') # Compile is used for grouping, we can have the part of a group
```

```python
match_group = re.search(phone_pattern,sen)
```

```python
match_group.group()
```

Exclusion
To exclude characters, we can use the ^ symbol in conjunction with a set of brackets []. Anything inside the brackets is excluded. For example:

```python
re.findall(r'[^\d]','There are 3 numbers 34 in 5 sentence')
```

```python
re.findall(r'[^\d]+','There are 3 numbers 34 in 5 sentence')
```

```python
test_phrase = 'This is a string! But it has punctuation. How can we remove it?'

re.findall(r'[^!.? ]',test_phrase)
```

```python
clean = ' '.join(re.findall(r'[^!.? ]+',test_phrase))
print(clean)
```

```python
text = 'Only find the hypen-words in this sentence. But you do not know how long-ish they are'
re.findall(r'[\w]+-[\w]+',text)
```

Parenthesis for Multiple Options
If we have multiple options for matching, we can use parenthesis to list out these options. For Example:

```python
text1='Would you like to have catfish?'
text2='DO you have a caterpillar'
text3='do you take a catnap?'

re.findall(r'cat(fish|erpillar|nap)',text3)
```
