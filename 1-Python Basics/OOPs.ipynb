```python
class Dog():

    # We can define a CLASS OBJECT ATTRIBUTE
    # NO NEED TO USE A SELF KEYWORD
    # THIS IS SAME FOR ANY INSTANCE OF THE CLASS

    species='Mammal'

    def __init__(self,mybreed,name,spots):
        # Attributes (are the characteristics of the Object)
        # We take in the argument (in this case, it is 'mybreed')
        # Self keyword represents the specific instance of the class and lets us use the variables,arguments and methods of the Object
        # Assign it using self.attribute_name (Attribute_name could be anything, but it is cleaner to have all the same name)
        self.breed=mybreed
        self.name=name

        # Expect a Boolean True/False
        self.spots=spots

    # OPERATIONS/METHODS --> ACTIONS
    # We can even pass outside arguments in the Bark class, here number is passed

    def Bark(self, number):
        print(f'{self.name} says Woof!! {number} number of times')
    
```

```python
my_Dog=Dog(mybreed='Huskie',name='Sammy',spots='No Spots')
```

```python
my_Dog.Bark(9)
```

```python
class Circle():

    # HERE LET'S DEFINE CLASS OBJECT ATTRIBUTE

    pi=3.14

    def __init__(self,radius=1):
        self.radius=radius
        self.area=radius*radius*self.pi # or Circle.pi

    def get_circum(self):

        return 2*Circle.pi*self.radius # HERE 'pi' NEEDS TO BE REFERENCED, EITHER EXPLICITLY BY WRITING CLASS NAME (CIRCLE.PI) OR by referencing the class attribute 'self.pi'
```

```python
cir=Circle(100) # cir=Circle() --> This is also right, when given this, it takes default radius value of 1
```

```python
cir.area
```

```python
cir.get_circum()
```

OOPS CONCEPTS --> POLYMORPHISM AND INHERITANCE 

```python
class Animal():

    def __init__(self):
        print('Animal class created')
    
    def who_am_i(self):
        print('I am an Animal')

    def eat(self):
        print('I am eating')
```

```python
class Dog(Animal):   # THIS IS INHERITANCE

    def __init__(self):
        Animal.__init__(self)
        print('Dog Created')
    
    def who_am_i(self):
        print('I am a dog and I am a mammal')
```

```python
mydog=Dog()
```

```python
mydog.eat()
```

```python
mydog.who_am_i()
```

# ABSTRACT CLASSES
# IS A CLASS, WHICH NEVER EXPECTS TO BE INITIATED/CREATE AN INSTANCE OF THE CLASS, IT IS DESIGNED ONLY AS A BASE CLASS

```python
class Dog():

    def __init__(self,name):
        self.name=name

    def speak(self):
        return self.name+ ' says Woof'

class Cat():

    def __init__(self,name):
        self.name=name

    
    def speak(self):
        return self.name+ ' says Meow!!'
       
```

```python
nido=Dog('Tommy')
fiko=Cat('Pussy')
```

```python
nido.speak()
```

```python
fiko.speak()
```

```python
for pet in [nido,fiko]:
    print(pet.speak())
```

```python
def pet_speak(pet):
    print(pet.speak())

pet_speak(nido)
pet_speak(fiko)
```

```python
class Book():

    def __init__(self,title, author,pages):
        self.title=title
        self.author=author
        self.pages=pages

    def __str__(self): # This __str__ is a special function which lets you print
        return 'The title of the book is {}'.format(self.title) # Make sure to use only return NOT PRINT
    
    def __len__(self):
        return 'Number of pages in the book is {}'.format(self.pages)
    
    def __del__(self):
        return 'The book is deleted'
```

```python
book = Book('Python Rocks!!','Jose',450)
```

```python
print(book)
```

```python
book.__len__()
```

```python
del book
```

```python
book.__len__()
```

```python
class Line():

    def __init__(self,coor1,coor2):
        self.coor1=coor1
        self.coor2=coor2 #self.x1=coor1[0] self.x2=coor1[1]

    def distance(self):
        x1,y1=self.coor1
        x2,y2=self.coor2 # either this approach or |
        return (y2-y1)/(x2-x1)
```

```python
coord1=(3,2)
coord2=(10,20)

li= Line(coord1,coord2)
```

```python
li.distance()
```

```python
class Cylinder():

    def __init__(self,radius,height):
        self.radius=radius
        self.height=height
    
    def volu(self):
        return (self.height)*(3.14)*(self.radius)**2
```

```python
c= Cylinder(34,78)
```

```python
c.volu()
```

```python
class Bank_Account():

    def __init__(self,owner,balance):
        self.owner=owner
        self.balance=balance

    def Deposit(self, credit):        

        self.balance = self.balance+credit
        
        print('The account now has a balance of {}'.format(self.balance))
        
    def Withdraw(self,debit):

        print('The balance in your account is {}'.format(self.balance))

        if self.balance > debit:

            self.balance=self.balance-debit

            print('you can withdraw the amount entered, now the balance in your account is {}'.format(self.balance))         
                     
        else:
            print('Insufficient balance, you cannot withdraw {}, your balance is {}'.format(debit,self.balance))
                  

```

```python
bank_details = Bank_Account('Karthik',100)
```

```python
bank_details.Deposit(100)
```

```python
bank_details.Withdraw(300)
```

```python
bank_details.Deposit(400)
```

```python
bank_details.Withdraw(300)
```
