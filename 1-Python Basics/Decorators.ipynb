# Suppose a function is created and you wish to add more functionality to it

you have two options:

Add that functionality to the old code
or else, Create a brand new function that contains Old code and then add new code to that function

But what if you want to delete that extra functionality

you would need to delete it manually or make sure to have the old function.

Is there a better way like an ON/OFF Switch to quickly add that functionality ?

Python has decorators that allow you to tack on extra functionality to an already existing function
They use @ operator and are then placed on top of the original function

@ Some_decorator

def simp_func():
    # Do something
    return something


```python
def fun():
    return 'Hello!'

```

```python
def hello(name='Jose'):

    print('This is hello() function')

    def greet():
        return '\t This is a greet() function inside hello function'

    def Welcome():
        return '\t This is a welcome() function inside hello function'

    print('I am going to return a function')

    if name == 'Jose':
        return greet
    else:
        return Welcome
```

```python
my_new_func=hello('Jose')
```

```python
my_new_func()
```

```python
print(my_new_func())
```

```python
def new_decorator(fu):

    def wrap_fun():
        print('Some code would go here before executing this function')
        fu()
        print('some code would go here after executing this function')
    return wrap_fun


def some_newfun():
    print('This is some function in need of a decorator')
    
```

```python
some_newfun()
```

```python
some_newfun = new_decorator(some_newfun)
```

```python
some_newfun()
```

Now let's see how python uses it using an @ symbol

```python
@new_decorator
def some_nfun():
    print('This is some function in need of a decorator')
```

```python
some_nfun()
```
