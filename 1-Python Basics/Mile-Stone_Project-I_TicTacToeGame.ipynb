```python
import random 

def choose_first():
    if random.randint(0,1)==0:
        return 'player 2'
    else:
        return 'player 1'
```

```python
def space_check(board,position):

    return board[position]==' '
```

```python
def full_boardcheck(board):

    for i in range(1,10):

        if space_check(board,i):
            return False
    return True


```

```python
def player_choice(board):
    position=0

    while position not in [1,2,3,4,5,6,7,8,9] or not space_check(board,position):

        position=int(input('Choose a position in the board from 1 to 9: '))

    return position
```

```python
def replay():

    return ("Do you wish to play this game again? (Y|N)").lower().startswith('y')
```
