```python
start = datetime.datetime(2006,1,1)
end = datetime.datetime(2016,1,1)
```

```python
import yfinance as yf

BAC = yf.download("BAC",start=start,end=end)
```

```python
BAC.head()
```

```python
# CitiGroup
C = yf.download("C",start, end)

# Goldman Sachs
GS = yf.download("GS",start, end)

# JPMorgan Chase
JPM = yf.download("JPM",start, end)

# Morgan Stanley
MS = yf.download("MS",start, end)

# Wells Fargo
WFC = yf.download("WFC",start, end)

```

```python
tickers = ['BAC', 'C', 'GS', 'JPM', 'MS', 'WFC']
```

** Use pd.concat to concatenate the bank dataframes together to a single data frame called bank_stocks. Set the keys argument equal to the tickers list. Also pay attention to what axis you concatenate on.**

```python
bank_stocks = pd.concat([BAC, C, GS, JPM,MS, WFC],axis = 1, keys = tickers)
```

```python
print(bank_stocks.columns)
print(bank_stocks.columns.nlevels)
```

** Set the column name levels (this is filled out for you):**
