# Groupby

The groupby method allows you to group rows of data together and call aggregate functions

```python
import pandas as pd
import numpy as np
```

```python
data = {'Company':['GOOG','GOOG','META','META','CAPG','CAPG'],
        'Sales': [100,123,453,467,178,200],
         'Empname':['KK','Ajay','Suhu','Bhavana','Aswath','Vikram'] }
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
df.groupby('Company')
```

```python
by_comp = df.groupby("Company")
```

```python
by_comp['Sales'].mean()
```

```python
by_comp['Sales'].std()
```

```python
by_comp.min()
```

```python
by_comp.max()
```

```python
by_comp.count()
```

```python
by_comp.describe()
```

```python
by_comp.describe().transpose() # just an example that you can transpose, the above one is more readable.
```

```python
by_comp.describe()
```

# Merging, Joining, and Concatenating

There are 3 main ways of combining DataFrames together: Merging, Joining and Concatenating. In this lecture we will discuss these 3 methods with examples.

____

```python
data_MJC1 = {'A': ['A0','A1','A2','A3'],
             'B': ['B0','B1','B2','B3'],
             'C': ['C0','C1','C2','C3'],
             'D': ['D0','D1','D2','D3']}
```

```python
df1 = pd.DataFrame(data_MJC1, index='0 1 2 3'.split())
```

```python
df1
```

```python
data_MJC2= {'A': ['A4','A5','A6','A7'],
             'B': ['B4','B5','B6','B7'],
             'C': ['C4','C5','C6','C7'],
             'D': ['D4','D5','D6','D7']}
```

```python
df2 = pd.DataFrame(data_MJC2, index='4 5 6 7'.split())
df3 = pd.DataFrame(data_MJC3, index= '8 9 10 11'.split())
```

```python
df2
```

## Example DataFrames

```python
left = pd.DataFrame({'Key':['Key0','Key1','Key2','Key3'],
                     'A': ['A0','A1','A2','A3'],
                     'B': ['B0','B1','B2','B3']
                     })

right = pd.DataFrame({'Key':['Key0','Key1','Key2','Key3'],
                      'C': ['C0','C1','C2','C3'],
                      'D': ['D0','D1','D2','D3']})
```

```python
left
```

```python
right
```

```python
pd.merge(left,right, how='inner', on='Key')
```

## Joining
Joining is a convenient method for combining the columns of two potentially differently-indexed DataFrames into a single result DataFrame.

```python
left1 = pd.DataFrame({'A': ['A0', 'A1', 'A2'],
                     'B': ['B0', 'B1', 'B2']},
                      index=['K0', 'K1', 'K2']) 

right1 = pd.DataFrame({'C': ['C0', 'C2', 'C3'],
                    'D': ['D0', 'D2', 'D3']},
                      index=['K0', 'K2', 'K3'])
```

```python
left1.join(right1)
```

```python
left1.join(right1, how='outer')
```
