I took the values of col1, col2 as string values instead of integers, that is why I had to convert them to numeric by using the below function, that is why str, int comparision errors are coming

```python
df1= pd.to_numeric(df['col1'])>2 
```

```python
df2=pd.to_numeric(df['col2'])>444
```

```python
newdf = df[(df1)&(df2)]
```
