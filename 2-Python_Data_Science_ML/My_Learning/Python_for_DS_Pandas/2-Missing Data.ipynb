# Missing Data

Let's show a few convenient methods to deal with Missing Data in pandas:

```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({'A':[1,2,np.NaN],
                   'B':[5,np.NaN,np.NaN],
                   'C':[1,2,3]}, index='KK SUH BHA'.split())
```

```python
df
```

```python
df.dropna()
```

```python
df.dropna(axis=1)
```

```python
df.fillna(value='Fill Value')
```

```python
df['A'].fillna(value=df['A'].mean())
```
